# *Bar Chart showing the average salves per weekday AA*

```python
import matplotlib.pyplot as plt
import datetime
#This function calculates the average sales per weekday
# for a given DataFrame containing sales data.
def calculate_average_sales_per_day(df):
    # Convert 'Datum' to datetime format
    df['Datum'] = pd.to_datetime(df['Datum'])
    
    # Extract the day of the week from 'Datum'
    df['Wochentag'] = df['Datum'].dt.day_name()
    
    # Group by 'Wochentag' and calculate the average sales
    average_sales = df.groupby('Wochentag')['Umsatz'].mean().reset_index()
    
    # Sort the days of the week
    days_order = ['Montag', 'Dienstag', 'Mittwoch', 'Donnerstag', 'Freitag', 'Samstag', 'Sonntag']
    average_sales['Wochentag'] = pd.Categorical(average_sales['Wochentag'], categories=days_order, ordered=True)
    
    return average_sales.sort_values('Wochentag')

df_average_sales_per_day = calculate_average_sales_per_day(df_umsatzdaten_gekuerzt)

# Einfache Daten
kategorien = ['Montag', 'Dienstag', 'Mittwoch', 'Donnerstag', 'Freitag', 'Samstag', 'Sonntag']
werte = df_average_sales_per_day['Umsatz']

# Einfaches Balkendiagramm erstellen
plt.figure(figsize=(8, 5))
plt.bar(kategorien, werte)

# Titel und Beschriftungen
plt.title('Einfaches Balkendiagramm')
plt.xlabel('Kategorien')
plt.ylabel('Umsatz')

# Diagramm anzeigen
plt.show()
```
