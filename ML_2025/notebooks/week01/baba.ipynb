```python
import pandas as pd

# Load the data.txt file into a DataFrame with proper encoding and delimiter
file_path = 'C:\\Users\\Beyer\\Documents\\VSCode_Workspace\\datalumina-project-template\\data\\raw\\data.txt'
data = pd.read_csv(file_path, encoding='utf-8', sep=',', on_bad_lines='skip')  # Comma-separated file

# Preview the first few rows of the data
print(data.head())

# Display basic information about the dataset
print(data.info())

# Handle missing values (example: fill with 0)
data = data.fillna(0)

# Save the cleaned data to a new CSV file (optional)
data.to_csv('cleaned_data.csv', index=False)
```

```python
def calculate_mean_temperature(data):
    """
    Calculate the mean of the 'Temperatur' column in the dataset.

    Parameters:
        data (pd.DataFrame): The DataFrame containing the data.

    Returns:
        float: The mean temperature.
    """
    if 'Temperatur' in data.columns:
        return data['Temperatur'].mean()
    else:
        raise ValueError("The 'Temperatur' column is not present in the dataset.")

# Example usage
mean_temperature = calculate_mean_temperature(data)
print(f"The mean temperature is: {mean_temperature}.2f")
```
